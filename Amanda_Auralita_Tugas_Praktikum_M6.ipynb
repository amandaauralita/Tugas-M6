```python
#NOMER 1
usia=int(input("Usia: "))
departemen=input("Nama Departemen: ")
#poin A dan B(usia pendfatar minimal 19 tahun dan departemen mengandung kata statistika)
syarat_asisten=(usia>=19) and ("Statistika" in departemen)
print(f"Apakah pendaftar asisten laboratorium memenuhi syarat?:",syarat_asisten)
```

```python
#NOMER 2
data_pelanggan=("Amanda","Premium")
berat_kardus=[20,30,50,25] #dalam satuan kg
(nama,member)=data_pelanggan #membongkar tuple
total_berat=sum(berat_kardus)
jalur_vip=("Premiun" in data_pelanggan) or (total_berat>100)
#output
print("Status member:",member)
print("Total berat seluruh barang:",total_berat,"kg")
print(f"Apakah pengiriman masuk jalur VIP?",jalur_vip)
```
